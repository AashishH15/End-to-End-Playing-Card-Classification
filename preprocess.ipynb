```python
import torch
import torchvision
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder

import os
```

```python
data_dir = os.getcwd() + "/cards/dataset_filtered/train"
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
])
data = ImageFolder(data_dir, transform=transform)


image, label = data[-1000]
image
```

```python
suits = [0] * 4
numbers = [0] * 13

suits[0]='clubs'
suits[1]='diamonds'
suits[2]='hearts'
suits[3]='spades'

numbers[0]='ace'
numbers[1]='eight'
numbers[2]='five'
numbers[3]='four'
numbers[4]='jack'
numbers[5]='king'
numbers[6]='nine'
numbers[7]='queen'
numbers[8]='seven'
numbers[9]='six'
numbers[10]='ten'
numbers[11]='three'
numbers[12]='two'

label_to_name = {}
count = 0
for number in numbers:
  for suit in suits:
    label_to_name[count] = f"{number} of {suit}"
    count += 1

label_to_name
```
